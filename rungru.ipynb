```python
import dash
from dash import dcc, html, Input, Output
import dash_bootstrap_components as dbc
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np

# 1. Extract and format the data from the trained pipeline
# We need to re-run prepare_data to grab X and metadata since they aren't stored as class attributes
print("Extracting sequences for the dashboard...")
X_dash, y_dash, meta_dash = pipeline.prepare_data(pcdata)
df_meta = pd.DataFrame(meta_dash)
df_meta['array_index'] = df_meta.index # Map metadata back to the X tensor

# 2. Initialize the Dash App
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.FLATLY])

# 3. Define the App Layout
app.layout = dbc.Container([
    html.H2("MCRPC Patient Follow-up & Risk Trajectory Dashboard", className="mt-4 mb-4 text-center"),
    
    dbc.Row([
        # Left Column: Patient & Visit Selection
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5("Patient Selection", className="card-title"),
                    dcc.Dropdown(
                        id='patient-dropdown',
                        options=[{'label': f"Patient {pid}", 'value': pid} for pid in df_meta['pid'].unique()],
                        value=df_meta['pid'].iloc[0],
                        clearable=False
                    ),
                    html.Br(),
                    html.H5("Follow-up Visit (Study Day)", className="card-title"),
                    dcc.Dropdown(
                        id='visit-dropdown',
                        clearable=False
                    )
                ])
            ], className="mb-4"),
            
            # Risk Score Display
            dbc.Card([
                dbc.CardBody([
                    html.H5(f"{pipeline.horizon}-Day Mortality Risk Score", className="card-title text-center"),
                    dcc.Graph(id='risk-gauge', config={'displayModeBar': False})
                ])
            ])
        ], md=4),
        
        # Right Column: Variable Readings & Trajectories
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5("Current Variable Readings (Unscaled)", className="card-title"),
                    html.Div(id='current-readings-table'),
                ])
            ], className="mb-4"),
            
            dbc.Card([
                dbc.CardBody([
                    html.H5("Variable Trajectories (Last 5 Visits)", className="card-title"),
                    dcc.Graph(id='trajectory-plots')
                ])
            ])
        ], md=8)
    ])
], fluid=True)

# 4. Callbacks for Interactivity

# Update Visit Dropdown based on Selected Patient
@app.callback(
    Output('visit-dropdown', 'options'),
    Output('visit-dropdown', 'value'),
    Input('patient-dropdown', 'value')
)
def update_visit_dropdown(selected_pid):
    patient_visits = df_meta[df_meta['pid'] == selected_pid].sort_values('current_vday')
    options = [{'label': f"Day {int(day)}", 'value': day} for day in patient_visits['current_vday']]
    return options, options[0]['value'] # Default to the first visit

# Update Dashboard Content based on Patient and Visit
@app.callback(
    Output('risk-gauge', 'figure'),
    Output('current-readings-table', 'children'),
    Output('trajectory-plots', 'figure'),
    Input('patient-dropdown', 'value'),
    Input('visit-dropdown', 'value')
)
def update_dashboard(pid, visit_day):
    # Find the specific index for this patient-visit
    visit_data = df_meta[(df_meta['pid'] == pid) & (df_meta['current_vday'] == visit_day)]
    if visit_data.empty:
        return go.Figure(), "No data available.", go.Figure()
    
    idx = visit_data['array_index'].iloc[0]
    
    # 1. Get the sequence and unscale it
    # X_dash[idx] shape: (12, num_features)
    seq_scaled = X_dash[idx]
    seq_unscaled = pipeline.scaler.inverse_transform(seq_scaled)
    
    # 2. Calculate Risk Score
    # We pass a batch of 1 to the model: shape (1, 12, num_features)
    raw_pred = pipeline.model.predict(X_dash[idx:idx+1], verbose=0)[:, -1, 0]
    risk_score = pipeline.final_scaler.transform(raw_pred.reshape(-1, 1)).flatten()[0]
    risk_score = np.clip(risk_score, 0, 1)
    
    # Generate Gauge Chart
    gauge_fig = go.Figure(go.Indicator(
        mode="gauge+number",
        value=risk_score,
        domain={'x': [0, 1], 'y': [0, 1]},
        gauge={
            'axis': {'range': [0, 1]},
            'bar': {'color': "rgba(44, 62, 80, 0.8)"},
            'steps': [
                {'range': [0, 0.3], 'color': "rgba(46, 204, 113, 0.3)"},
                {'range': [0.3, 0.7], 'color': "rgba(241, 196, 15, 0.3)"},
                {'range': [0.7, 1.0], 'color': "rgba(231, 76, 60, 0.3)"}
            ],
            'threshold': {
                'line': {'color': "red", 'width': 4},
                'thickness': 0.75,
                'value': 0.5 # You can replace this with your optimized 'best_t' threshold
            }
        }
    ))
    gauge_fig.update_layout(height=250, margin=dict(l=20, r=20, t=30, b=20))
    
    # 3. Current Variable Readings (Table)
    # The last time step in the sequence represents the current visit
    current_readings = seq_unscaled[-1]
    
    table_header = [html.Thead(html.Tr([html.Th("Variable"), html.Th("Reading")]))]
    table_rows = []
    for col_name, val in zip(pipeline.feature_cols, current_readings):
        table_rows.append(html.Tr([html.Td(col_name), html.Td(f"{val:.2f}")]))
    table_body = [html.Tbody(table_rows)]
    readings_table = dbc.Table(table_header + table_body, bordered=True, hover=True, size="sm")
    
    # 4. Trajectory Plots (Last 5 visits)
    # Filter out padding (represented by extreme negative values or logic) 
    # and just take the last 5 steps of the sequence.
    lookback = min(5, pipeline.sequence_length)
    recent_seq = seq_unscaled[-lookback:]
    visit_indices = list(range(-lookback + 1, 1)) # e.g., [-4, -3, -2, -1, 0] (0 is current)
    
    rows = (len(pipeline.feature_cols) + 1) // 2
    traj_fig = make_subplots(rows=rows, cols=2, subplot_titles=pipeline.feature_cols, vertical_spacing=0.1)
    
    for i, col_name in enumerate(pipeline.feature_cols):
        r = (i // 2) + 1
        c = (i % 2) + 1
        traj_fig.add_trace(
            go.Scatter(
                x=visit_indices, 
                y=recent_seq[:, i], 
                mode='lines+markers',
                name=col_name,
                line=dict(color='#2980b9', width=2),
                marker=dict(size=8)
            ),
            row=r, col=c
        )
        # Highlight current visit
        traj_fig.add_trace(
            go.Scatter(
                x=[0], 
                y=[recent_seq[-1, i]], 
                mode='markers',
                marker=dict(color='#e74c3c', size=10, symbol='star'),
                showlegend=False
            ),
            row=r, col=c
        )
        
    traj_fig.update_layout(
        height=200 * rows, 
        showlegend=False, 
        margin=dict(l=20, r=20, t=40, b=20),
        xaxis_title="Visits Relative to Current (0)",
        plot_bgcolor="rgba(240, 240, 240, 0.5)"
    )
    # Update all x-axes to have integer ticks
    traj_fig.update_xaxes(tickmode='array', tickvals=visit_indices)

    return gauge_fig, readings_table, traj_fig

# 5. Run the Server
if __name__ == '__main__':
    # Use standard port 8051. debug=True allows hot-reloading if you change code.
    app.run(debug=True, port=8051)
```

```python
import dash
from dash import dcc, html, Input, Output, State
import dash_bootstrap_components as dbc
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np
from sklearn.metrics import confusion_matrix, roc_curve, auc, precision_recall_curve, average_precision_score, balanced_accuracy_score
from lifelines.utils import concordance_index

# --- 1. Global Data & Metrics Calculation ---
X_dash, y_dash, meta_dash = pipeline.prepare_data(pcdata)
df_meta = pd.DataFrame(meta_dash)
df_meta['array_index'] = df_meta.index

best_t = getattr(pipeline, 'best_t', 0.5)
raw_preds_all = pipeline.model.predict(X_dash, verbose=0)[:, -1, 0]
risks_all = np.clip(pipeline.final_scaler.transform(raw_preds_all.reshape(-1, 1)).flatten(), 0, 1)
y_pred_all = (risks_all >= best_t).astype(int)

# Global Metrics for the Report Section
tte = np.array([m['tte'] for m in meta_dash])
events = np.array([m['event'] for m in meta_dash])
c_index = concordance_index(tte, 1 - risks_all, events)

# Visit-Level Confusion Matrix
cm_v = confusion_matrix(y_dash, y_pred_all)
tn_v, fp_v, fn_v, tp_v = cm_v.ravel()

# Patient-Level Confusion Matrix
df_eval = df_meta.copy()
df_eval['y_pred'], df_eval['y_true'] = y_pred_all, y_dash
patient_summary = df_eval.groupby('pid').agg({'y_true': 'max', 'y_pred': 'max'})
cm_p = confusion_matrix(patient_summary['y_true'], patient_summary['y_pred'])
tn_p, fp_p, fn_p, tp_p = cm_p.ravel()

# --- 2. Dash App Setup ---
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.FLATLY])

app.layout = dbc.Container([
    html.H2("MCRPC Mortality Risk Analytics Dashboard", className="mt-4 mb-4 text-center"),

    # --- SECTION: GLOBAL PERFORMANCE & SENSITIVITY ---
    dbc.Card([
        dbc.CardHeader(html.H4("Model Performance & Risk Sensitivity", className="mb-0")),
        dbc.CardBody([
            dbc.Row([
                # Comprehensive Metrics Column
                dbc.Col([
                    html.H5("Comprehensive Metrics", className="text-primary"),
                    html.P(f"Global C-Index: {c_index:.4f}", className="mb-1"),
                    html.P(f"Threshold: {best_t:.3f}", className="small text-muted"),
                    html.Hr(),
                    html.B("Visit-Level"),
                    html.Table([
                        html.Tr([html.Td("Sens (Recall)"), html.Td(f"{tp_v/(tp_v+fn_v):.3f}")]),
                        html.Tr([html.Td("Spec"), html.Td(f"{tn_v/(tn_v+fp_v):.3f}")]),
                        html.Tr([html.Td("PPV (Prec)"), html.Td(f"{tp_v/(tp_v+fp_v):.3f}")]),
                        html.Tr([html.Td("Bal. Acc"), html.Td(f"{balanced_accuracy_score(y_dash, y_pred_all):.3f}")]),
                    ], className="table table-sm"),
                    html.B("Patient-Level"),
                    html.Table([
                        html.Tr([html.Td("Sens"), html.Td(f"{tp_p/(tp_p+fn_p):.3f}")]),
                        html.Tr([html.Td("Spec"), html.Td(f"{tn_p/(tn_p+fp_p):.3f}")]),
                    ], className="table table-sm"),
                ], md=3),

                # Performance Grid (plot_performance_grid)
                dbc.Col([
                    dcc.Graph(id='performance-grid')
                ], md=9),
            ]),
            html.Hr(),
            # Risk Sensitivity Marginals (Two Rows)
            html.H5("Risk Sensitivity Marginals (Delta Risk per Feature Shift)", className="mb-3"),
            dcc.Graph(id='sensitivity-marginals')
        ])
    ], className="mb-4 shadow-sm"),

    # --- SECTION: PATIENT LOOKUP ---
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5("Patient Selection"),
                    dcc.Dropdown(
                        id='patient-dropdown',
                        options=[{'label': f"Patient {pid}", 'value': pid} for pid in df_meta['pid'].unique()],
                        value=df_meta['pid'].unique()[0],
                        clearable=False
                    ),
                    html.Br(),
                    html.H5("Visit Day"),
                    dcc.Dropdown(id='visit-dropdown', clearable=False)
                ])
            ], className="mb-4"),
            dbc.Card([
                dbc.CardBody([
                    html.H5("Mortality Risk", className="text-center"),
                    dcc.Graph(id='risk-gauge'),
                    html.Div(id='alarm-status-container', className="text-center")
                ])
            ])
        ], md=4),
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5("Predictor Trajectories (Last 5 Visits)"),
                    dcc.Graph(id='trajectory-plots')
                ])
            ])
        ], md=8)
    ])
], fluid=True)

# --- 3. Callbacks ---

@app.callback(
    [Output('performance-grid', 'figure'),
     Output('sensitivity-marginals', 'figure')],
    Input('patient-dropdown', 'value') # Trigger once
)
def render_global_analytics(_):
    # 1. Performance Grid
    grid = make_subplots(rows=1, cols=4, subplot_titles=("ROC AUC", "Prec-Recall", "Sens/Spec", "PPV/NPV"))
    
    fpr, tpr, _ = roc_curve(y_dash, risks_all)
    grid.add_trace(go.Scatter(x=fpr, y=tpr, name="ROC", fill='tozeroy'), row=1, col=1)
    
    prec, rec, _ = precision_recall_curve(y_dash, risks_all)
    grid.add_trace(go.Scatter(x=rec, y=prec, name="PR"), row=1, col=2)
    
    thresh_ax = np.linspace(0, 1, 50)
    s_l, sp_l, p_l, n_l = [], [], [], []
    for t in thresh_ax:
        p_temp = (risks_all >= t).astype(int)
        tn, fp, fn, tp = confusion_matrix(y_dash, p_temp).ravel()
        s_l.append(tp/(tp+fn)); sp_l.append(tn/(tn+fp))
        p_l.append(tp/(tp+fp) if (tp+fp)>0 else 0); n_l.append(tn/(tn+fn) if (tn+fn)>0 else 1)
        
    grid.add_trace(go.Scatter(x=thresh_ax, y=s_l, name="Sens"), row=1, col=3)
    grid.add_trace(go.Scatter(x=thresh_ax, y=sp_l, name="Spec"), row=1, col=3)
    grid.add_trace(go.Scatter(x=thresh_ax, y=p_l, name="PPV"), row=1, col=4)
    grid.add_trace(go.Scatter(x=thresh_ax, y=n_l, name="NPV"), row=1, col=4)
    grid.update_layout(height=350, showlegend=False, margin=dict(t=30, b=10))

    # 2. Risk Sensitivity (8 Plots in 2 Rows)
    sens_fig = make_subplots(rows=2, cols=4, subplot_titles=pipeline.feature_cols)
    sample_idx = np.random.choice(X_dash.shape[0], min(100, len(X_dash)), replace=False)
    X_s = X_dash[sample_idx]
    base_p = pipeline.model.predict(X_s, verbose=0)[:, -1, 0]
    offsets = np.linspace(-1.5, 1.5, 7)

    for i, col in enumerate(pipeline.feature_cols):
        r, c = (i // 4) + 1, (i % 4) + 1
        deltas = []
        for off in offsets:
            X_mod = X_s.copy()
            X_mod[:, -1, i] += off
            new_p = pipeline.model.predict(X_mod, verbose=0)[:, -1, 0]
            deltas.append(np.mean(new_p - base_p))
        sens_fig.add_trace(go.Scatter(x=offsets, y=deltas, mode='lines+markers', line_color='#2c3e50'), row=r, col=c)
        sens_fig.add_hline(y=0, line_dash="dash", line_color="red", row=r, col=c)

    sens_fig.update_layout(height=500, showlegend=False, margin=dict(t=40, b=20))
    return grid, sens_fig

@app.callback(
    [Output('visit-dropdown', 'options'), Output('visit-dropdown', 'value')],
    Input('patient-dropdown', 'value')
)
def sync_visits(pid):
    days = df_meta[df_meta['pid'] == pid].sort_values('current_vday')['current_vday']
    return [{'label': f"Day {int(d)}", 'value': d} for d in days], days.iloc[0]

@app.callback(
    [Output('risk-gauge', 'figure'), Output('alarm-status-container', 'children'), Output('trajectory-plots', 'figure')],
    [Input('patient-dropdown', 'value'), Input('visit-dropdown', 'value')]
)
def update_view(pid, vday):
    idx = df_meta[(df_meta['pid'] == pid) & (df_meta['current_vday'] == vday)]['array_index'].iloc[0]
    raw_p = pipeline.model.predict(X_dash[idx:idx+1], verbose=0)[:, -1, 0]
    risk = np.clip(pipeline.final_scaler.transform(raw_p.reshape(-1, 1)).flatten()[0], 0, 1)

    gauge = go.Figure(go.Indicator(mode="gauge+number", value=risk, 
        gauge={'axis':{'range':[0,1]}, 'threshold':{'line':{'color':"red",'width':4},'value':best_t}}))
    gauge.update_layout(height=250, margin=dict(t=30, b=10))

    is_alarm = risk >= best_t
    alarm = dbc.Badge("CRITICAL ALARM" if is_alarm else "STABLE", color="danger" if is_alarm else "success", style={"fontSize": "1.2rem"})

    # Trajectories
    unscaled = pipeline.scaler.inverse_transform(X_dash[idx])
    traj = make_subplots(rows=2, cols=4, subplot_titles=pipeline.feature_cols)
    for i, col in enumerate(pipeline.feature_cols):
        traj.add_trace(go.Scatter(y=unscaled[-5:, i], mode='lines+markers'), row=(i//4)+1, col=(i%4)+1)
    traj.update_layout(height=450, showlegend=False, margin=dict(t=40, b=20))

    return gauge, alarm, traj

if __name__ == '__main__':
    app.run(debug=True, port=8050)
```
