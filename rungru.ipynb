```python
import dash
from dash import dcc, html, Input, Output
import dash_bootstrap_components as dbc
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np

# 1. Extract and format the data from the trained pipeline
# We need to re-run prepare_data to grab X and metadata since they aren't stored as class attributes
print("Extracting sequences for the dashboard...")
X_dash, y_dash, meta_dash = pipeline.prepare_data(pcdata)
df_meta = pd.DataFrame(meta_dash)
df_meta['array_index'] = df_meta.index # Map metadata back to the X tensor

# 2. Initialize the Dash App
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.FLATLY])

# 3. Define the App Layout
app.layout = dbc.Container([
    html.H2("MCRPC Patient Follow-up & Risk Trajectory Dashboard", className="mt-4 mb-4 text-center"),
    
    dbc.Row([
        # Left Column: Patient & Visit Selection
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5("Patient Selection", className="card-title"),
                    dcc.Dropdown(
                        id='patient-dropdown',
                        options=[{'label': f"Patient {pid}", 'value': pid} for pid in df_meta['pid'].unique()],
                        value=df_meta['pid'].iloc[0],
                        clearable=False
                    ),
                    html.Br(),
                    html.H5("Follow-up Visit (Study Day)", className="card-title"),
                    dcc.Dropdown(
                        id='visit-dropdown',
                        clearable=False
                    )
                ])
            ], className="mb-4"),
            
            # Risk Score Display
            dbc.Card([
                dbc.CardBody([
                    html.H5(f"{pipeline.horizon}-Day Mortality Risk Score", className="card-title text-center"),
                    dcc.Graph(id='risk-gauge', config={'displayModeBar': False})
                ])
            ])
        ], md=4),
        
        # Right Column: Variable Readings & Trajectories
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5("Current Variable Readings (Unscaled)", className="card-title"),
                    html.Div(id='current-readings-table'),
                ])
            ], className="mb-4"),
            
            dbc.Card([
                dbc.CardBody([
                    html.H5("Variable Trajectories (Last 5 Visits)", className="card-title"),
                    dcc.Graph(id='trajectory-plots')
                ])
            ])
        ], md=8)
    ])
], fluid=True)

# 4. Callbacks for Interactivity

# Update Visit Dropdown based on Selected Patient
@app.callback(
    Output('visit-dropdown', 'options'),
    Output('visit-dropdown', 'value'),
    Input('patient-dropdown', 'value')
)
def update_visit_dropdown(selected_pid):
    patient_visits = df_meta[df_meta['pid'] == selected_pid].sort_values('current_vday')
    options = [{'label': f"Day {int(day)}", 'value': day} for day in patient_visits['current_vday']]
    return options, options[0]['value'] # Default to the first visit

# Update Dashboard Content based on Patient and Visit
@app.callback(
    Output('risk-gauge', 'figure'),
    Output('current-readings-table', 'children'),
    Output('trajectory-plots', 'figure'),
    Input('patient-dropdown', 'value'),
    Input('visit-dropdown', 'value')
)
def update_dashboard(pid, visit_day):
    # Find the specific index for this patient-visit
    visit_data = df_meta[(df_meta['pid'] == pid) & (df_meta['current_vday'] == visit_day)]
    if visit_data.empty:
        return go.Figure(), "No data available.", go.Figure()
    
    idx = visit_data['array_index'].iloc[0]
    
    # 1. Get the sequence and unscale it
    # X_dash[idx] shape: (12, num_features)
    seq_scaled = X_dash[idx]
    seq_unscaled = pipeline.scaler.inverse_transform(seq_scaled)
    
    # 2. Calculate Risk Score
    # We pass a batch of 1 to the model: shape (1, 12, num_features)
    raw_pred = pipeline.model.predict(X_dash[idx:idx+1], verbose=0)[:, -1, 0]
    risk_score = pipeline.final_scaler.transform(raw_pred.reshape(-1, 1)).flatten()[0]
    risk_score = np.clip(risk_score, 0, 1)
    
    # Generate Gauge Chart
    gauge_fig = go.Figure(go.Indicator(
        mode="gauge+number",
        value=risk_score,
        domain={'x': [0, 1], 'y': [0, 1]},
        gauge={
            'axis': {'range': [0, 1]},
            'bar': {'color': "rgba(44, 62, 80, 0.8)"},
            'steps': [
                {'range': [0, 0.3], 'color': "rgba(46, 204, 113, 0.3)"},
                {'range': [0.3, 0.7], 'color': "rgba(241, 196, 15, 0.3)"},
                {'range': [0.7, 1.0], 'color': "rgba(231, 76, 60, 0.3)"}
            ],
            'threshold': {
                'line': {'color': "red", 'width': 4},
                'thickness': 0.75,
                'value': 0.5 # You can replace this with your optimized 'best_t' threshold
            }
        }
    ))
    gauge_fig.update_layout(height=250, margin=dict(l=20, r=20, t=30, b=20))
    
    # 3. Current Variable Readings (Table)
    # The last time step in the sequence represents the current visit
    current_readings = seq_unscaled[-1]
    
    table_header = [html.Thead(html.Tr([html.Th("Variable"), html.Th("Reading")]))]
    table_rows = []
    for col_name, val in zip(pipeline.feature_cols, current_readings):
        table_rows.append(html.Tr([html.Td(col_name), html.Td(f"{val:.2f}")]))
    table_body = [html.Tbody(table_rows)]
    readings_table = dbc.Table(table_header + table_body, bordered=True, hover=True, size="sm")
    
    # 4. Trajectory Plots (Last 5 visits)
    # Filter out padding (represented by extreme negative values or logic) 
    # and just take the last 5 steps of the sequence.
    lookback = min(5, pipeline.sequence_length)
    recent_seq = seq_unscaled[-lookback:]
    visit_indices = list(range(-lookback + 1, 1)) # e.g., [-4, -3, -2, -1, 0] (0 is current)
    
    rows = (len(pipeline.feature_cols) + 1) // 2
    traj_fig = make_subplots(rows=rows, cols=2, subplot_titles=pipeline.feature_cols, vertical_spacing=0.1)
    
    for i, col_name in enumerate(pipeline.feature_cols):
        r = (i // 2) + 1
        c = (i % 2) + 1
        traj_fig.add_trace(
            go.Scatter(
                x=visit_indices, 
                y=recent_seq[:, i], 
                mode='lines+markers',
                name=col_name,
                line=dict(color='#2980b9', width=2),
                marker=dict(size=8)
            ),
            row=r, col=c
        )
        # Highlight current visit
        traj_fig.add_trace(
            go.Scatter(
                x=[0], 
                y=[recent_seq[-1, i]], 
                mode='markers',
                marker=dict(color='#e74c3c', size=10, symbol='star'),
                showlegend=False
            ),
            row=r, col=c
        )
        
    traj_fig.update_layout(
        height=200 * rows, 
        showlegend=False, 
        margin=dict(l=20, r=20, t=40, b=20),
        xaxis_title="Visits Relative to Current (0)",
        plot_bgcolor="rgba(240, 240, 240, 0.5)"
    )
    # Update all x-axes to have integer ticks
    traj_fig.update_xaxes(tickmode='array', tickvals=visit_indices)

    return gauge_fig, readings_table, traj_fig

# 5. Run the Server
if __name__ == '__main__':
    # Use standard port 8050. debug=True allows hot-reloading if you change code.
    app.run(debug=True, port=8050)
```
